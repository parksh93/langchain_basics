```python
from langchain_ollama import ChatOllama

llm = ChatOllama(model="llama3.2:1b")

llm.invoke("What is the capital of Korea?")
```

```python
from langchain_core.prompts import PromptTemplate

prompt_template = PromptTemplate(
    template="What is the capital of {country}?",
    input_variables=["country"],
)

prompt = prompt_template.invoke({"country" : "Korea"})

llm.invoke(prompt)
```

```python
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage

# 예제를 넣어준다고 생각하면 됨
messagelist = [
    SystemMessage(content="You are a helpful assistent!"),
    HumanMessage(content="What is the capital of Korea?"),
    AIMessage(content="The capital of Korea is Seoul"),
    HumanMessage(content="What is the capital of German")
]

llm.invoke(messagelist)
```

```python
from langchain_core.prompts import ChatPromptTemplate

chat_prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant!"),
    ("human", "What is the capital of {country}"),
])

chat_prompt = chat_prompt_template.invoke({"country" : "Korea"})

llm.invoke(chat_prompt)
```
