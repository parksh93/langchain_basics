```python
from langchain_ollama import ChatOllama

llm = ChatOllama(
    model="llama3.2:1b",
    temperature=0,
)
```

```python
# 음식 이름 찾기
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate

food_prompt = PromptTemplate(
    template="""what is one of the most popula food in {country}? please return the name of food only""",
    input_variables=['country']
)

output = StrOutputParser()

food_chain = food_prompt | llm | output
```

```python
food_chain.invoke({'country': 'Korea'})
```

```python
# 음식 레시피 찾기
from langchain_core.prompts import ChatPromptTemplate

recipe_prompt = ChatPromptTemplate.from_messages({
    ('system', 'Provide the recipe of the food that the user wants. Please return the recipe only'),
    ('human', 'Can you give me the recipe for making {food}')
})

recipe_chain = recipe_prompt | llm | output
```

```python
recipe_chain.invoke({'food', 'Bibimbap'})
```

```python
full_chain = {'food' :food_chain} | recipe_chain

full_chain.invoke({'country' : 'Korea'})
```
